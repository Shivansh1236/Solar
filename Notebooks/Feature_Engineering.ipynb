```python
import pandas as pd
import numpy as np

gen = pd.read_csv("../data/cleaned/Plant_1_Generation_Cleaned.csv")
weather = pd.read_csv("../data/cleaned/Plant_1_Weather_Cleaned.csv")

gen["DATE_TIME"] = pd.to_datetime(gen["DATE_TIME"])
weather["DATE_TIME"] = pd.to_datetime(weather["DATE_TIME"])

merged = pd.merge(
    gen,
    weather,
    on=["DATE_TIME", "PLANT_ID"],
    how="inner"
)

merged.head()
```

```python
merged["HOUR"] = merged["DATE_TIME"].dt.hour
merged["DAY"] = merged["DATE_TIME"].dt.day
merged["MONTH"] = merged["DATE_TIME"].dt.month
merged["WEEKDAY"] = merged["DATE_TIME"].dt.weekday

merged["IS_WEEKEND"] = (
    merged["WEEKDAY"] >= 5
).astype(int)

merged.head()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

merged["SOURCE_KEY_x"] = encoder.fit_transform(
    merged["SOURCE_KEY_x"]
)

merged.head()
```

```python
merged = merged.drop(columns=["DATE_TIME"])

merged.info()
```

```python
merged.to_csv(
    "../data/cleaned/Plant_1_Feature_Engineered.csv",
    index=False
)
```
